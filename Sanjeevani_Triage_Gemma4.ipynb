```python
def triage_assistant(symptom_text):
    if "सिरदर्द" in symptom_text or "थकान" in symptom_text:
        return "Green (Mild): Likely stress or dehydration."
    elif "पेट दर्द" in symptom_text or "उल्टी" in symptom_text:
        return "Yellow (Doctor Visit): Possible infection, needs evaluation."
    elif "सांस" in symptom_text or "सीने में दर्द" in symptom_text:
        return "Red (Emergency): Warning signs of heart/respiratory issue."
    else:
        return "Please consult a doctor for clarification."

# Example run
print(triage_assistant("मुझे हल्का सिरदर्द और थकान है"))

```
